```python
import duckdb
import datasets

ds = datasets.load_dataset('RevolutionCrossroads/si_us_revolutionary_era_collections',
split= 'train'
)

arrow_table = ds.data.table
connection = duckdb.connect()

continental_currency = connection.sql("""
SELECT
title , date, mediaURLs
FROM
  arrow_table
WHERE
  title ILIKE '%CONTINENTAL%'
  AND title ILIKE '%DOLLAR%'
  ORDER BY date
  """)


colonial_currency = connection.sql("""
SELECT
title , date, indexed_places, mediaURLs
FROM
  arrow_table
WHERE
indexed_topics::VARCHAR ILIKE '%NNC COLONIAL CURRENCY%'
AND indexed_topics::VARCHAR NOT ILIKE '%CONTINENTAL%'
ORDER BY date
""")




```
