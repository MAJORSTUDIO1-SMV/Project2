```python
dataset = load_dataset(
    "RevolutionCrossroads/nara_revolutionary_war_pension_files",
    split="train",
    streaming=True,
)
```

```python


from transformers import pipeline
import torch
from tqdm import tqdm

device = 0 if torch.cuda.is_available() else -1
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=0
)

results = []
count = 0

for row in dataset:
    text = row.get('RevX_OCR_text')
    if text:
        new_text = classifier(
            text,
            candidate_labels=['pension payments', 'bounty land warrant', 'other']
        )

        if new_text['labels'][0] in ['pension payments', 'bounty land warrant'] and new_text['scores'][0] >= 0.75:
            results.append(new_text)
            print(new_text)





```
